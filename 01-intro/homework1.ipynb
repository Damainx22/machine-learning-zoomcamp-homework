## Q7. Sum of weights

```python
df_Asia = df.query("origin == 'Asia'")
df_Asia.head(7)
```

```python
X_df = df_Asia[['vehicle_weight', 'model_year']].head(7)
X_df
```

```python
X = X_df.to_numpy()
X
```

```python
XTX = np.dot(X.T, X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = np.dot((np.dot(XTX_inv, X.T)), y)
w
```

```python
w.sum()
```
